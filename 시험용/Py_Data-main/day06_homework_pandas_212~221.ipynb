## 결측치

#### 데이터 불러오기

```python
import pandas as pd
url = 'https://raw.githubusercontent.com/sehakflower/data/main/titanic.csv'
t_df = pd.read_csv(url,sep='\t')
t_df.info()
```

#### columns 소문자로 변경

```python
new_columns = ['passengerId','survived','pclass','name','sex','age','sibsp','parch','ticket','fare','cabin','embarked']

t_df.columns = new_columns
t_df.columns
```

#### isna 함수

```python
t_df['age'][t_df['age'].isna()]#isna로 결측치인 NaN추출
```

#### isnull 함수

```python
t_df['age'].isnull()#age columns에서 null이면 True 값이 있으면 False 반환
```

```python
t_df['age'][t_df['age'].isnull()]#isnull로 결측치인 NaN추출
```

#### sum 함수

```python
t_df['age'][t_df['age'].isnull()].isnull().sum()#age columns에 결측치가 얼마나 있는지 체크(True = 1,False=0인 성질 이용)
```

#### notnull함수

```python
t_df['age'].notnull()# 결측치가 없으면 True반환
```

```python
t_df['age'][t_df['age'].notnull()]# NaN이 아닌 값만 표시
```

```python
t_df.loc[t_df['age'].notnull(),['age','name']]# age에서 NaN이 아닌 값을 표시하고 age와 name columns을 같이 표시
```

```python
t_df.loc[t_df['age'].isnull(),['age','name']]# age에서 NaN인 값을 표시하고 age와 name columns을 같이 표시
```

## 결측치 처리 방법

#### 결측치를 다른 문자로 변경

```python
t_df.head()#앞 데이터 5개 확인
```

```python
t_df['cabin'] = t_df['cabin'].fillna('N')# cabin의 결측치를 다 'N'으로 변경
```

```python
t_df['cabin'].value_counts() #cabin열에서 개별값 모두 확인
```

```python
t_df['cabin'][0]#cabin의 0번째 인덱스 확인
```

```python
t_df['cabin'] = t_df['cabin'].apply(lambda x : x[0])#데이터의 첫문자만 남김
t_df['cabin']
```

```python
t_df.info()# 대략 정보 확인
```

#### 결측치를 평균 값으로 변경

```python
t_df_copy = t_df.copy()# 데이터 프레임 복제
t_df_copy
```

```python
t_df_copy['age'].isnull() #결측치 확인
```

```python
t_df_copy['age'] = t_df_copy['age'].fillna(t_df['age'].mean()) # 결측치를 평균 값으로 변경

t_df_copy['age']
```
