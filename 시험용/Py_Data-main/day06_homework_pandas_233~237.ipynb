## 판다스의 통계 함수

```python
import pandas as pd
url = 'https://raw.githubusercontent.com/sehakflower/data/main/titanic.csv'
t_df = pd.read_csv(url,sep='\t')
```

#### 데이터의 통계치 확인하기

```python
t_df.describe #통계치 확인
```

#### sum 함수

```python
t_df.sum()
```

#### count 함수

```python
t_df.count()
```

#### max 함수

```python
t_df.max()
```

#### min 함수

```python
t_df.min()
```

#### mean 함수

```python
t_df.mean(numeric_only=True)
```

#### median 함수

```python
t_df.median(numeric_only=True)
```

#### mode 함수

```python
t_df.mode()
```

#### mad 함수

```python
t_df.mad()
```

```python
num_t_df = t_df.select_dtypes(include='number')
mad = (num_t_df - num_t_df.mean()).abs().mean()
print(mad)
```

#### var 함수

```python
t_df.var(numeric_only=True)
```

#### std 함수

```python
t_df.std(numeric_only=True)
```

#### skew 함수

```python
t_df.skew(numeric_only=True)
```

#### kurt 함수

```python
t_df.kurt(numeric_only=True)
```
