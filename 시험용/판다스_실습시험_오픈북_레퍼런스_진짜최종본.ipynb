# 6. Boolean Indexing — 조건으로 행 추출

**중요**
- `and` 대신 `&`
- `or` 대신 `|`
- 조건 하나하나를 반드시 `( )`로 묶기

### 자주 쓰는 형태
```python
df[df['점수'] >= 80]

df[(df['점수'] >= 80) & (df['지역'] == '서울')]

df[(df['지역'] == '서울') | (df['지역'] == '부산')]

df[df['지역'].isin(['서울', '부산'])]
```


````python
print('점수 80 이상')
display(df[df['점수'] >= 80])

print('점수 80 이상 AND 서울')
display(df[(df['점수'] >= 80) & (df['지역'] == '서울')])

print('서울 OR 부산')
display(df[(df['지역'] == '서울') | (df['지역'] == '부산')])

#지역 : 서울 또는 부산
print('isin 사용')
display(df[df['지역'].isin(['서울', '부산'])])
````

## 6-A. Boolean Indexing — 변형 대응 집중

### 1. 단일 조건
```python
df[df['점수'] >= 80]
```

### 2. AND
```python
df[(df['점수'] >= 80) & (df['등급'] == 'A')]
```

### 3. OR
```python
df[(df['지역'] == '서울') | (df['지역'] == '부산')]
```

### 4. 범위 조건
```python
df[(df['나이'] >= 20) & (df['나이'] < 30)]
```

### 5. 특정 값 여러 개
```python
df[df['지역'].isin(['서울','부산'])]
```

### 6. 특정 값 제외
```python
df[~df['지역'].isin(['서울','부산'])]
```

### 7. 조건에 맞는 특정 열만
```python
df.loc[df['점수'] >= 80, ['이름','점수']]
```

### 8. 문자열 조건
```python
df[df['이름'].str.contains('Mr', na=False)]
```

### 9. 결측치 조건
```python
df[df['점수'].isnull()]
```

### 10. 결측치가 아닌 조건
```python
df[df['점수'].notnull()]
```

**문제 문장 해석**
- "~이고" → `&`
- "~이거나" → `|`
- "~중 하나" → `isin()`
- "~가 아닌" → `~`

# 8. value_counts() — 값별 개수

**문제 문장**
- 지역별 인원 수를 구하시오.
- 등급별 데이터 개수를 구하시오.
- 특정 컬럼에서 각 값이 몇 번 나오는지 구하시오.

기본:
```python
df['컬럼'].value_counts()
```

비율:
```python
df['컬럼'].value_counts(normalize=True)
```


# 9. groupby() — 그룹별 집계

수업 실습의 대표 형태:

# 17-A. apply() + lambda — 실제 데이터 변환 실습

`map()`과 비슷하게 **각 값에 함수를 적용**할 때 사용한다.

```python
titanic_df['cabin'] = titanic_df['cabin'].fillna('N')

titanic_df['cabin'] = titanic_df['cabin'].apply(

)
```

예:
```text
C85       → C
C123      → C
F G73     → F
NaN → N   → N
```

### 범용형

```python
df['새열'] = df['기존열'].apply(lambda x: 변환식)
```

예: 문자열의 첫 글자
```python
df['첫글자'] = df['문자열열'].apply(lambda x: x[0])
```

### map과 apply
이 시험 범위에서는 둘 다 **각 값에 함수를 적용하는 도구**로 우선 이해해도 된다.

```python
df['새열'] = df['열'].map(lambda x: ...)
```
