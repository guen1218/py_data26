```python
import pandas as pd
# 주어진 파일에 있는 데이터를 데이터프레임으로 로딩
datafile = 'midterm_exam_data.csv'
exam_df = pd.read_csv(datafile)

# 앞에서 3개의 데이터 확인 
exam_df.head(3)
```

```python
# 뒤에서 7개의 데이터 확인
exam_df.tail(7)
```

```python
# 데이터 개수 확인
len(exam_df)
```

```python
# 데이터의 구성 정보 확인 (인덱스, 칼럼, 결측치, 데이터 수, 데이터 타입 등을 한번에 확인)
exam_df.info()
```

```python
# 데이터의 기본 통계치 보기
exam_df.describe()
```

```python
# 열 이름 변경하기 : 모두 소문자로 변경 (str.lower 함수 활용)
exam_df = exam_df.rename(columns=str.lower)
```

```python
# backup_df로 exam_df 저장 (deep copy)
backup_df = exam_df.copy()
```

```python
# 행의 순서를 거꾸로 출력
exam_df[::-1]
```

```python
# 컬럼의 수 출력
exam_df.shape[1]
```

```python
# 'name', 'age', 'sex' 컬럼 출력
exam_df[['name', 'age', 'sex']]
```

```python
# 'embarked' 컬럼에 결측치가 있는 행의 전체 데이터 출력
exam_df[exam_df['embarked'].isna()]
```

```python
# 'embarked' 컬럼에 결측치가 있는 행 삭제
exam_df.dropna(subset=['embarked'], inplace=True)
```

```python
# 'embarked' 컬럼에 결측치가 남아 있는지 확인
exam_df[exam_df['embarked'].isna()]
```

```python
# 'age' 컬럼의 결측치 갯수 구하기
exam_df['age'].isnull().sum()
```

```python
# 'age' 컬럼에 결측치가 있는 행의 마지막 컬럼 데이터 출력 (컬럼명을 사용하지 말 것)
exam_df[exam_df['age'].isna()][-1:]
```

```python
# 'age' 컬럼의 결측치를 'age' 컬럼의 최대값 채우기
exam_df['age'] = exam_df['age'].fillna(exam_df['age'].max())
```

```python
# 'age' 컬럼의 결측치 갯수 구하기
exam_df['age'].isnull().sum()
```

```python
# 'sex' 컬럼의 결측치가 있는 행의 'passengerid'부터 'age' 컬럼까지 출력
exam_df.loc[exam_df['sex'].isna(), 'passengerid':'age']
```

```python
# 'sex' 컬럼에 결측치가 있는 데이터 삭제
exam_df.dropna(subset=['sex'], inplace=True)
```

```python
# 'sex'의 컬럼 값과 빈도수 확인
exam_df['sex'].value_counts(sort=True)
```

```python
# 'sex' 컬럼의 값을 첫글자의 대문자로 수정 dk tlqkfdk
exam_df['sex'] = exam_df['sex'].apply(lambda x: x[0].upper() + x[1:])

# 인덱스 10부터 15까지의 데이터 확인
exam_df[10:16]
```

```python
# 전체 데이터가 같은 중복행의 수
exam_df.duplicated().sum()
```

```python
# 전체 데이터가 같은 중복행 출력 (첫번째 데이터도 출력)
exam_df[exam_df.duplicated(keep=False)]
```

```python
# 전체 데이터가 같은 중복행 삭제
exam_df = exam_df.drop_duplicates()
```

```python
# 전체 데이터가 같은 중복행이 있는지 확인 (중복 데이터의 수)
exam_df.duplicated().sum()
```

```python
# 전처리가 완료된 exam_df를 저장 : OOO을 본인 한글이름으로 수정
result_file = 'midterm_exam_result_이근휘.csv'
exam_df.to_csv(result_file)
```

```python
# 'pclass'별 생존자의 수 확인
exam_df[['pclass','survived']].value_counts()
exam_df.groupby('pclass')['survived'].sum()
```

```python
# 'pclass', 'sex'의 조합에 대한 생존 여부의 평균를 표로 출력
pd.pivot_table(
    exam_df,
    values='survived',
    columns=['pclass', 'sex'],
    aggfunc='mean'
)
```
