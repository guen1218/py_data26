```python
def getIndex(num_list, target):
    return num_list.index(target)
```

```python
def getMax(num_list) :
    return max(num_list)
```

```python
def getMin(num_list) :
    return min(num_list)
```

```python
def countGT(num_list, target):
    count=0
    for i in num_list:
        if int(i)>target:
            count+=1
    return count
```

```python
def sumList(num_list) :
    return sum(num_list)
```

```python
def swapList(num_list):
    return num_list.reverse()
```

```python
number_list = [23, 45, 27, 11, 25, 65, 78]

print (getIndex(number_list,25))
print (getMax(number_list))
print (getMin(number_list))
print (countGT(number_list,42))
print (sumList(number_list))
swapList(number_list)
print (number_list)
```

```python
def getIndex(number_list, target):
    return number_list.index(target)
```

```python
def getMax(number_list):
    return max(number_list)
```

```python
def getMin(number_list):
    return min(number_list)
```

```python
def countGT(number_list, target):
    return sum(1 for i in number_list if i > target)
```

```python
def sumList(number_list):
    return sum(number_list)
```

```python
def swapList(number_list):
    number_list.reverse()
```

```python
number_list = [23, 45, 27, 11, 25, 65, 78]

print(getIndex(number_list,25))
print(getMax(number_list))
print(getMin(number_list))
print(countGT(number_list,42))
print(sumList(number_list))
swapList(number_list)
print(number_list)
```
